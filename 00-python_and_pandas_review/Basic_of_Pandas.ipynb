```python
import pandas as pd

# 1. Creating DataFrames
df_1 = pd.DataFrame(
                  # specify values for each column - column-wise
                  {
                    'id' : [1,2,3],
                    'name' : ['Alex', 'George', 'Anna'],
                    'age': [23, 25, 36],
                    'gender': ['M', 'M', 'F']
                  },
                  index=[1,2,3]
                  )
```

```python
df_1.head()
```

```python
df_2 = pd.DataFrame(
                  # specify values for each row - row-wise
                  [
                    [1,'Alex',23,'M'],
                    [2, 'George', 25, 'M'],
                    [3, 'Anna', 36, 'F']
                  ],
                  index=[1,2,3],
                  columns=['id','name','age','gender']
                  )
```

```python
df_2.head()
```

```python
pd.melt(df_2)
```

```python
# joining two dfs'
pd.concat([df_1,df_2])
```

```python
# order rows by values of a column
df_1=df_1.sort_values('name')
```

```python
df_1
```

```python
df_1=df_1.sort_values('age',ascending=False)
```

```python
df_1
```
