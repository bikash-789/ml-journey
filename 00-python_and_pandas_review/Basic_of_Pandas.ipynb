```python
# rename column attendance --> attendance_percentage
df.rename(columns={'Attendance':'Attendance_Percentage'}, inplace=True)
```

```python
# create a new column 'Result' based on score
df["Result"] = "Needs Improvement"

df.loc[df["Score"] >= 70, "Result"] = "Average"
df.loc[df["Score"] >= 80, "Result"] = "Good"
df.loc[df["Score"] >= 90, "Result"] = "Excellent"
```

```python
# Perform this in one statement:
'''
    1. Keep only students with Score > 80.
    2. Rename Name to Full_Name.
    3. Select only Full_Name, Department, and Score.
    4. Sort by Score descending. '''

df_1 = (
    df[df['Score'] > 80]
      .rename(columns={'Name':'Full_Name'})
      .loc[:,['Full_Name','Department', 'Score']]
      .sort_values('Score', ascending=False)
)
df_1
```
