## <I>Phase-II : Data Cleaning

```python
orders_df.info()
```

```python
# since we have some columns which doesn't have appropriate dtypes, we'll try to convert those

orders_df['order_purchase_timestamp'] = pd.to_datetime(orders_df['order_purchase_timestamp'])
orders_df['order_approved_at'] = pd.to_datetime(orders_df['order_approved_at'])
orders_df['order_delivered_carrier_date'] = pd.to_datetime(orders_df['order_delivered_carrier_date'])
orders_df['order_delivered_customer_date'] = pd.to_datetime(orders_df['order_delivered_customer_date'])
orders_df['order_estimated_delivery_date'] = pd.to_datetime(orders_df['order_estimated_delivery_date'])
```

```python
order_items_df['shipping_limit_date'] = pd.to_datetime(order_items_df['shipping_limit_date'])
```

```python
order_reviews_df['review_creation_date'] = pd.to_datetime(order_reviews_df['review_creation_date'])
order_reviews_df['review_answer_timestamp'] = pd.to_datetime(order_reviews_df['review_answer_timestamp'])
```

```python
order_reviews_df.info()
```

```python
order_reviews_df.head()
```

```python
# duplicate check
dfs = {
    'customers': customers_df,
    'geolocation': geolocation_df,
    'order_items': order_items_df,
    'order_payments': order_payments_df,
    'order_reviews': order_reviews_df,
    'orders': orders_df,
    'products': products_df,
    'sellers': sellers_df,
    'category_name_translation': category_name_translation_df
}

def check_for_duplicates(dfs):
    for name, df in dfs.items():
        print(f"{name}: {df.duplicated().sum()} duplicates")

check_for_duplicates(dfs)
```

```python
dfs['geolocation'].info()
```

```python
geolocation_df.head()
```

```python
# let's check which rows are duplicated
geolocation_df[geolocation_df.duplicated(keep=False)]
```

```python
geolocation_df[geolocation_df.duplicated()]
```

```python
# drop duplicates since it have repeated values across rows
geolocation_df = geolocation_df.drop_duplicates(keep='first')
```

```python
geolocation_df[geolocation_df.duplicated()]
```

```python
geolocation_df.duplicated().sum()
```

```python
# if we don't update our dataset (geolocation_df) with the updated dataset (cleaned)
# then it won't show us 0 duplicates, the object still holds previous dataset
dfs['geolocation'] = geolocation_df # updating with new one
check_for_duplicates(dfs)
```
