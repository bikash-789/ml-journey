```python
# order analysis

orders = orders_df[['order_id', 'customer_id', 'order_status', 'order_purchase_timestamp', 
        'order_delivered_customer_date', 'order_estimated_delivery_date']] \
        .merge(order_reviews_df[['order_id', 'review_score']], on='order_id', how='inner')

orders.head()
```

```python
# product analysis

products = order_items_df[['order_id', 'product_id', 'price', 
            'freight_value', 'seller_id']] \
            .merge(products_df[['product_category_name', 'product_id']], on='product_id', how='inner') \
            .merge(category_name_translation_df[['product_category_name', 'product_category_name_english']], on='product_category_name', how='left')

products.head()
```

```python
# customer analysis
print(customers_df.duplicated(subset=['customer_id']).sum())
print(customers_df.duplicated(subset=['customer_unique_id']).sum())
```

```python
customer_summary = (order_items_df[['order_id', 'price']]\
            .merge(orders[['order_id', 'customer_id', 'order_purchase_timestamp']], on='order_id', how='left') \
            .groupby('customer_id') \
            .agg(
                order_count=('order_id', 'count'),
                total_spending=('price', 'sum'),
                last_purchase_date=('order_purchase_timestamp', 'max')
            ).reset_index())

customer_summary['order_count'] = customer_summary['order_count'].astype('Int64')

customers = (
    customers_df[['customer_id', 'customer_state']]
    .merge(
        customer_summary,
        on='customer_id',
        how='left'
    )
)
customers.query('order_count > 1')
```

### Utils

```python
# util function to convert figures to human readable format
def formatNumber(n):
    if n >= 1_000_000_000:
        return f"{n / 1_000_000_000:.1f}B"
    elif n >= 1_000_000:
        return f"{n / 1_000_000:.1f}M"
    elif n >= 1_000:
        return f"{n / 1_000:.1f}K"
    else:
        return f"{n:.0f}"
    
def printnl():
    print('\n')
```

### Business Questions

#### 1. What is the total revenue?

```python
# 1. What is the total revenue?
print(f'Total revenue = R${formatNumber(customers["total_spending"].sum())}')
printnl()
```

#### 2. What is the average order value?

```python
# 2. What is the average order value?
products['order_item_value'] = products['price'] + products['freight_value']
order_values = products \
                    .groupby('order_id') \
                    .agg(order_value=('order_item_value', 'sum')) \
                    .reset_index()
average_order_value = order_values['order_value'].mean()
print(f'Average order value: R${formatNumber(average_order_value)}')
printnl()
```

#### 3. What are the top 10 product categories by revenue?

```python
# 3. What are the top 10 product categories by revenue ?
category_revenue = (
    products.groupby('product_category_name_english')
    [['price', 'freight_value']]
    .sum()
    .assign(revenue=lambda df: df['price'] + df['freight_value'])
    [['revenue']]
    .sort_values('revenue', ascending=False)
)
category_revenue.head(10)
```

#### 4. Which categories have the highest average order value?

```python
# 4. Which categories have the highest average order value?
order_values_per_category = products \
                        [['product_category_name_english', 'order_id', 'order_item_value']] \
                        .groupby(['product_category_name_english', 'order_id']) \
                        .agg(order_value=('order_item_value', 'sum')) \
                        .reset_index()

products_avg_order_value = order_values_per_category \
                            .groupby('product_category_name_english') \
                            .agg( avg_order_value=('order_value', 'mean')) \
                            .sort_values('avg_order_value', ascending=False) \
                            .reset_index()

print('Categories with highest average order value:\n')
print(products_avg_order_value.head())
printnl()
```

#### 5. Which categories generate the most orders?

```python
# 5. Which categories generate the most orders?
order_count_per_category = products[['product_category_name_english', 'order_id']] \
                            .groupby('product_category_name_english') \
                            .agg(order_count=('order_id', 'nunique')) \
                            .sort_values('order_count', ascending=False)

order_count_per_category.head()
```

#### 6. How many unique customers are there?

```python
# 6. How many unique customers are there?
unique_customers_cnt = customers['customer_id'].nunique()
print(f'Unique customers count is {unique_customers_cnt}')
```

#### 7. How many customers made 1, 2, or 3+ orders?

```python
# 7. How many customers made 1, 2, or 3+ orders?
import numpy as np
bins = [0, 1, 2, np.inf] # 1 interval (discrete)
labels = ['1 order', '2 orders', '3+ orders']

customers['order_bucket'] = pd.cut(customers['order_count'], bins=bins, labels=labels)

bucket_counts = customers['order_bucket'].value_counts()
print(bucket_counts)
```
